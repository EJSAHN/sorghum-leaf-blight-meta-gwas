This notebook is a thin wrapper around `src/run_final_pipeline.py`. Raw inputs are not distributed with the repository. Set the project root and GEMMA executable before running the command cell.


```python
from pathlib import Path
import os
import subprocess
import sys

REPOSITORY_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
PROJECT_ROOT = Path(os.environ.get('LEAF_BLIGHT_PROJECT_ROOT', '../leaf_blight_analysis')).resolve()
RUNNER = REPOSITORY_ROOT / 'src' / 'run_final_pipeline.py'
CONFIG = REPOSITORY_ROOT / 'config' / 'final_config.json'
GEMMA_EXECUTABLE = os.environ.get('GEMMA_EXECUTABLE', '/path/to/gemma-0.98.5')
```

```python
command = [
    sys.executable, str(RUNNER),
    '--project-root', str(PROJECT_ROOT),
    '--stage', 'all',
    '--config', str(CONFIG),
    '--gemma-wsl', GEMMA_EXECUTABLE,
]
print(' '.join(command))
# subprocess.run(command, check=True)
```
